## Segmento 2: Que es una tupla?

Una tupla es una coleccion de datos parecida a una lista, pero con una diferencia clave:
**su estructura no cambia facilmente**.

Se escribe con parentesis, aunque Python tambien puede crearla sin ellos:

```python
coordenada = (10, 20)
registro = "Ana", 15, "Python"
```

Las tuplas son utiles cuando el orden importa y queremos conservarlo.


````python
coordenada = (10, 20)
print("Coordenada completa:", coordenada)
print("Valor en x:", coordenada[0])
print("Valor en y:", coordenada[1])
print("Cantidad de datos:", len(coordenada))
````

## Segmento 3: Diferencia entre lista y tupla

Una lista puede modificarse con facilidad.
Una tupla, en cambio, suele usarse para datos mas estables, como:

- Un horario fijo.
- Una fecha.
- Un resultado con varios valores.
- Un registro corto que no queremos reorganizar.

Esa idea ayuda a que nuestro programa sea mas claro.


````python
lista_colores = ["rojo", "azul", "verde"]
tupla_colores = ("rojo", "azul", "verde")

lista_colores[1] = "amarillo"
print("Lista modificada:", lista_colores)
print("Tupla original:", tupla_colores)
````

## Segmento 4: Desempaquetar una tupla

Una ventaja muy practica es que podemos guardar cada valor de la tupla en una variable distinta.
A eso se le llama **desempaquetado**.

Es como abrir una caja y repartir lo que contiene.


````python
persona = ("Luis", 16, "Ciudad de Mexico")
nombre, edad, ciudad = persona

print("Nombre:", nombre)
print("Edad:", edad)
print("Ciudad:", ciudad)
````

## Segmento 5: Funciones que devuelven varios datos

Una funcion puede regresar mas de un valor.
Cuando eso pasa, Python realmente devuelve una tupla.

Esto nos permite resolver tareas en un solo paso y luego separar los resultados.


````python
def analizar_precio(precio):
    impuesto = precio * 0.16
    total = precio + impuesto
    return impuesto, total

resultado = analizar_precio(250)
print("Resultado completo:", resultado)

impuesto, total = resultado
print("Impuesto:", impuesto)
print("Total:", total)
````

## Segmento 6: Que es una excepcion?

Una excepcion es un error que aparece mientras el programa se esta ejecutando.

Por ejemplo:

- Escribir texto cuando se esperaba un numero.
- Dividir entre cero.
- Pedir una posicion que no existe.

Si no manejamos ese error, el programa puede detenerse.
Si lo manejamos bien, podemos mostrar un mensaje claro y continuar.

## Segmento 7: Manejo basico con `try` y `except`

El bloque `try` contiene el codigo que podria fallar.
El bloque `except` indica que haremos si ocurre un error.

La idea no es ocultar los problemas, sino responder a ellos de una forma util.


````python
texto = "45"

try:
    numero = int(texto)
    print("Conversion correcta:", numero)
except ValueError:
    print("No se pudo convertir el texto en numero")
````

## Segmento 8: Cuando el dato es incorrecto

Veamos ahora un caso donde si ocurre el error.
Asi podemos comparar el comportamiento del programa.


````python
texto = "cuarenta y cinco"

try:
    numero = int(texto)
    print("Conversion correcta:", numero)
except ValueError:
    print("El texto no representa un numero entero")
````

## Segmento 9: Capturar distintos tipos de error

Un programa puede fallar por razones diferentes.
Por eso podemos usar mas de un `except`.

Cada bloque responde a un problema especifico.


````python
def dividir_valores(a, b):
    try:
        a = float(a)
        b = float(b)
        resultado = a / b
        print("Resultado:", resultado)
    except ValueError:
        print("Debes escribir numeros validos")
    except ZeroDivisionError:
        print("No se puede dividir entre cero")

dividir_valores("18", "3")
dividir_valores("18", "0")
dividir_valores("dieciocho", "3")
````

## Segmento 10: `else` y `finally`

Ademas de `try` y `except`, existen dos bloques opcionales:

- `else`: se ejecuta cuando no hubo error.
- `finally`: se ejecuta siempre.

Esto es util cuando queremos separar el caso exitoso del cierre del proceso.


````python
def mostrar_descuento(precio_texto):
    try:
        precio = float(precio_texto)
    except ValueError:
        print("El precio debe ser numerico")
    else:
        descuento = precio * 0.10
        print("Descuento:", descuento)
        print("Precio final:", precio - descuento)
    finally:
        print("Fin de la revision")

mostrar_descuento("350")
print("---")
mostrar_descuento("trescientos")
````

## Segmento 11: Crear nuestras propias excepciones

A veces el error no viene de Python, sino de una regla de nuestro programa.

Por ejemplo:

- La edad no puede ser negativa.
- La calificacion debe estar entre 0 y 10.
- No puede haber cupo negativo en un curso.

En esos casos podemos usar `raise` para lanzar una excepcion con un mensaje claro.


````python
def validar_calificacion(calificacion):
    if calificacion < 0 or calificacion > 10:
        raise ValueError("La calificacion debe estar entre 0 y 10")
    return calificacion

for dato in [8, 11]:
    try:
        print("Calificacion valida:", validar_calificacion(dato))
    except ValueError as error:
        print("Error detectado:", error)
````

## Segmento 12: Que es una clase?

Una clase es una forma de agrupar **datos** y **acciones** relacionadas.

Piensa en una clase como un molde.
A partir de ese molde podemos crear objetos.

Por ejemplo, si hacemos una clase `Mascota`, cada objeto puede guardar:

- nombre
- especie
- edad

Y ademas puede realizar acciones, como presentarse o cumplir anios.

## Segmento 13: Crear una clase sencilla

Para definir una clase usamos `class`.
Dentro escribimos funciones especiales llamadas metodos.

El metodo `__init__` se ejecuta cuando creamos un objeto.
Sirve para guardar los datos iniciales.


````python
class Mascota:
    def __init__(self, nombre, especie, edad):
        self.nombre = nombre
        self.especie = especie
        self.edad = edad

firulais = Mascota("Firulais", "perro", 4)

print(firulais.nombre)
print(firulais.especie)
print(firulais.edad)
````

## Segmento 14: Metodos de una clase

Un metodo es una funcion que pertenece a la clase.
Nos permite decir que puede hacer cada objeto.

La palabra `self` representa al objeto que esta trabajando en ese momento.


````python
class Mascota:
    def __init__(self, nombre, especie, edad):
        self.nombre = nombre
        self.especie = especie
        self.edad = edad

    def presentarse(self):
        print(f"Hola, soy {self.nombre} y soy un {self.especie}.")

    def cumplir_anios(self):
        self.edad += 1
        print(f"Ahora tengo {self.edad} anios.")

nina = Mascota("Nina", "gata", 2)
nina.presentarse()
nina.cumplir_anios()
````

## Segmento 15: Una clase tambien puede devolver tuplas

Las clases pueden trabajar con tuplas cuando necesitamos resumir varios datos.
Asi combinamos dos conceptos de la clase de hoy.


````python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def resumen(self):
        return self.nombre, self.precio

cuaderno = Producto("Cuaderno", 35)
print(cuaderno.resumen())
````

## Segmento 16: Clases + excepciones

Las clases tambien pueden proteger sus datos.
Si detectan un valor invalido, pueden lanzar una excepcion.

Eso hace que el objeto se comporte de forma mas segura.


````python
class CuentaAhorro:
    def __init__(self, titular, saldo):
        if saldo < 0:
            raise ValueError("El saldo inicial no puede ser negativo")
        self.titular = titular
        self.saldo = saldo

    def retirar(self, cantidad):
        if cantidad > self.saldo:
            raise ValueError("No hay saldo suficiente")
        self.saldo -= cantidad
        print(f"Retiro exitoso. Saldo actual: {self.saldo}")

try:
    cuenta = CuentaAhorro("Elena", 500)
    cuenta.retirar(200)
    cuenta.retirar(400)
except ValueError as error:
    print("Operacion cancelada:", error)
````

## Segmento 17: Mini proyecto final

Vamos a unir todo en un solo ejemplo.
Crearemos un pequeno sistema para registrar participantes en un club de retos.

Usaremos:

- `tuplas` para guardar cada inscripcion.
- `excepciones` para validar reglas.
- `clases` para organizar el programa.


````python
class Participante:
    def __init__(self, nombre, nivel):
        self.nombre = nombre
        self.nivel = nivel

    def ficha(self):
        return self.nombre, self.nivel


class ClubRetos:
    def __init__(self, cupo_maximo):
        if cupo_maximo <= 0:
            raise ValueError("El cupo maximo debe ser mayor que cero")
        self.cupo_maximo = cupo_maximo
        self.inscripciones = []

    def registrar(self, participante):
        if len(self.inscripciones) >= self.cupo_maximo:
            raise ValueError("Ya no hay lugares disponibles")
        self.inscripciones.append(participante.ficha())

    def mostrar_resumen(self):
        print("Resumen del club:")
        for nombre, nivel in self.inscripciones:
            print(f"- {nombre} | Nivel: {nivel}")
        print("Total inscritos:", len(self.inscripciones))


club = ClubRetos(2)

try:
    club.registrar(Participante("Ana", "inicial"))
    club.registrar(Participante("Luis", "intermedio"))
    club.registrar(Participante("Sara", "avanzado"))
except ValueError as error:
    print("No fue posible completar el registro:", error)

club.mostrar_resumen()
````

## Segmento 18: Leer el mini proyecto paso por paso

Observa estas decisiones:

- `Participante` guarda datos de una persona.
- `ficha()` devuelve una tupla con la informacion principal.
- `ClubRetos` controla el cupo y las inscripciones.
- Si se supera el limite, aparece una excepcion.

Este patron es muy comun en programas reales: cada clase tiene una responsabilidad clara.
